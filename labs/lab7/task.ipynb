## Задание 3 (варианты 1–15)

- для каждого управляющего кубита $k$ постройте управляемый оператор $M_b$ с $b = a^{2^k} \bmod N$ как `UnitaryGate` перестановки (по образцу функции `mod_mult_gate`); при $b = 1$ вентиль не нужен;

Перед запуском проверьте программно (`assert`), что каждый $M_b$ переводит $|x\rangle$ в $|bx \bmod N\rangle$ для всех $x < N$.

3. Запустите схему на `AerSimulator` (`shots=2000`). Для каждого результата $y$ вычислите $\theta = y/2^m$, дробь `Fraction(y, 2**m).limit_denominator(N)` и кандидата $r$.
